```python

import math


class Dual:
    def __init__(self, val, der=0.0):
        self.val = val
        self.der = der
    @staticmethod
    def _lift(x):
        return x if isinstance(x, Dual) else Dual(x, 0.0)


    def __add__(self, other):
        o = Dual._lift(other)
        return Dual(self.val + o.val, self.der + o.der)

    __radd__ = __add__

    def __sub__(self, other):
        o = Dual._lift(other)
        return Dual(self.val - o.val, self.der - o.der)

    def __rsub__(self, other):
        return Dual._lift(other) - self

    def __mul__(self, other):
        o = Dual._lift(other)

        return Dual(self.val * o.val, self.der * o.val + self.val * o.der)

    __rmul__ = __mul__

    def __truediv__(self, other):
        o = Dual._lift(other)

        return Dual(self.val / o.val,
                    (self.der * o.val - self.val * o.der) / (o.val ** 2))

    def __rtruediv__(self, other):
        return Dual._lift(other) / self

    def __pow__(self, n):

        return Dual(self.val ** n, n * self.val ** (n - 1) * self.der)

    def __neg__(self):
        return Dual(-self.val, -self.der)

    def __repr__(self):
        return f"Dual(val={self.val}, der={self.der})"


def exp(x):
    if isinstance(x, Dual):
        e = math.exp(x.val)
        return Dual(e, e * x.der)
    return math.exp(x)


def derivative(f, x):
    """Return (f(x), f'(x)) by seeding the input with derivative 1."""
    out = f(Dual(x, 1.0))
    out = Dual._lift(out)
    return out.val, out.der


if __name__ == "__main__":

    f = lambda x: x ** 3 - x ** 2
    val, der = derivative(f, 2.0)
    print(f"f(x)=x^3-x^2 at 2:  value={val}, derivative={der}")
    assert (val, der) == (4.0, 8.0)


    h = 1e-6
    fd = ((2 + h) ** 3 - (2 + h) ** 2 - ((2 - h) ** 3 - (2 - h) ** 2)) / (2 * h)
    print(f"finite difference: {fd:.8f}   dual number: {der:.8f}")


    g = lambda x: exp(-x * x) * x ** 2 + 3 / (1 + x)
    gv, gd = derivative(g, 1.3)

    x = 1.3
    exact = math.exp(-x * x) * (2 * x - 2 * x ** 3) - 3 / (1 + x) ** 2
    print(f"g'(1.3): dual={gd:.10f}  exact={exact:.10f}")
    assert abs(gd - exact) < 1e-12


    gap = lambda s: 1.5 - 2.0 * s + 0.5 * s ** 2
    target = 0.9
    loss = lambda s: (gap(s) - target) ** 2

    s = 0.0
    lr = 0.05
    for step in range(200):
        l, dl = derivative(loss, s)
        if step % 5 == 0 or l < 1e-10:
            print(f"step {step:3d}  strain={s:.6f}  loss={l:.3e}")
        if l < 1e-10:
            break
        s = s - lr * dl
    print(f"final strain={s:.6f}  gap={gap(s):.6f}  (target {target})")

    assert abs(gap(s) - target) < 1e-4
```
